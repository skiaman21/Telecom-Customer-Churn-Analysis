```python
#Standardize Column Names (snake_case for clean coding)
df.columns=df.columns.str.lower()
df.columns=df.columns.str.strip()
df.columns=df.columns.str.replace(' ','_')
```

# Handling missing values

```python
#offers
df['offer']=df['offer'].fillna('No offer')
```

```python
# Phone-related charges
df['avg_monthly_long_distance_charges']=df['avg_monthly_long_distance_charges'].fillna(0.0)
df['multiple_lines']=df['multiple_lines'].fillna('No Phone Service')
```

```python
#intenet related service
internet_services=[
    'internet_type','avg_monthly_gb_download','online_security',
    'online_backup','device_protection_plan','premium_tech_support',
    'streaming_tv','streaming_movies','streaming_music','unlimited_data'
]

for col in internet_services:
    df[col]=df[col].fillna('No Internet Service')
```

```python
#churn category and reason for stayed and joined customers
df['churn_category']=df['churn_category'].fillna('Not Churned')
df['churn_reason']=df['churn_reason'].fillna('Not Churned')
```

```python
df.isnull().sum()
```

```python
#create a new column tenure group
def tenure_group(months):
    if months<=6:
        return '0-6 Months'
    elif months<=12:
        return '6-12 Months'
    elif months<=24:
        return '1-2 Years'
    else:
        return '4+ Years'

df['tenure_group']=df['tenure_in_months'].apply(tenure_group)
```

```python
rev_q75 = df['total_revenue'].quantile(0.75)
rev_q25=df['total_revenue'].quantile(0.25)
```

```python
# create customer value segments(High Value, Mid Value, Low Value)
def assign_value_tier(revenue):
    if revenue>=rev_q75:
        return 'High Value (Top 25%)'
    elif revenue>=rev_q25:
        return 'Mid Value'
    else:
        return 'Low Value (Bottom 25%)'

df['customer_value_tier']=df['total_revenue'].apply(assign_value_tier)
```

```python
#At-Risk Customer Flag
#High Risk Customer=Month-to-Month Contract + No Premium Tech Support / No Online Security
df['is_high_risk']=np.where(
    (df['contract']=='Month-to-Month') &
    ((df['premium_tech_support']=='No')|(df['online_security']=='No')),'High Risk','Standard Risk')
```

```python
df['is_churned'] = (df['customer_status'] == 'Churned').astype(int)
```

```python
print("\n--- Data Cleaning Summary ---")
print(f"Total Rows Processed: {len(df)}")
print(f"Remaining Missing Values: {df.isnull().sum().sum()}")
print("\nCustomer Status Distribution:")
print(df['customer_status'].value_counts())
```

```python
#Exporting Cleaned Csv file
df.to_csv('telecom_churn_clean.csv',index=False)
```
